```python
df = pd.read_csv(f"jira_epic_tickets_extended_{EPIC_KEY}.csv")

with open(f"project_categories_{EPIC_KEY}.json", "r", encoding="utf-8") as f:
    category_data = json.load(f)

ticket_to_cat_subcat = {}

for category, subcats in category_data.items():
    for subcat, keys in subcats.items():
        for key in keys:
            ticket_to_cat_subcat[key] = (category, subcat)

df["Category"] = df["Key"].map(lambda k: ticket_to_cat_subcat.get(k, ("", ""))[0])
df["Subcategory"] = df["Key"].map(lambda k: ticket_to_cat_subcat.get(k, ("", ""))[1])

cols = ["Category", "Subcategory"] + [col for col in df.columns if col not in ["Category", "Subcategory"]]
df = df[cols]

df.to_csv("jira_epic_with_categories.csv", index=False)
print("✅ Saved as jira_epic_with_categories.csv")
```
